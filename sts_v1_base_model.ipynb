## Treino e Teste do Modelo - STS

```python
import torch
from types import SimpleNamespace
from sts_classifier import train_sts, test_sts, seed_everything
```

```python
def set_args_sts(option="finetune", lr=1e-5, epochs=5):
    args = SimpleNamespace(
        train="data/sts-train.csv",
        dev="data/sts-dev.csv",
        test="data/sts-test-student.csv",
        checkpoint_path="sts-checkpoint.pt",
        dev_out=f"predictions/{option}-sts-dev-output.csv",
        test_out=f"predictions/{option}-sts-test-output.csv",
        lr=lr,
        epochs=epochs,
        batch_size=8,
        hidden_dropout_prob=0.1,
        seed=11711,
        option=option,
        use_gpu=True if torch.cuda.is_available() else False
    )
    return args

seed_everything(11711)
```

## Pretrain

```python
# Treinamento Pretrain
args = set_args_sts(option="pretrain", lr=1e-3, epochs=10)
train_sts(args)

# Teste
test_sts(args)
```

## Finetune

```python
# Treinamento Finetune
args = set_args_sts(option="finetune", lr=1e-5, epochs=6)
train_sts(args)

# Teste
test_sts(args)
```
