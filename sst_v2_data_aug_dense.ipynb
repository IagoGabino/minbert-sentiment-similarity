## Treino e Teste do Modelo com melhorias e camada Densa - SST 

Adam - weight decay, Finetune com Smart Loss, Data Aug, Camada Densa antes da Classificação 

```python
import torch
import pandas as pd
import argparse
from sst_classifier import train, test, seed_everything
from types import SimpleNamespace
```

```python
# Configuração do treino e teste do modelo
def set_args(dataset, option, lr, epochs = 10):
    args = SimpleNamespace(
        filepath=f'{dataset}-classifier.pt',
        lr=lr,
        use_gpu=True if torch.cuda.is_available() else False,
        epochs=6,
        batch_size=8,
        hidden_dropout_prob=0.3,
        train=f'data/ids-{dataset}-train-augmented.csv',
        dev=f'data/ids-{dataset}-dev.csv',
        test=f'data/ids-{dataset}-test-student.csv',
        option=option,
        dev_out=f'predictions/{option}-{dataset}-dev-out.csv',
        test_out=f'predictions/{option}-{dataset}-test-out.csv'
    )
    return args
```

## Finetune

```python
# Definir argumentos
seed_everything(11711)
option = "finetune"
datasets = ["sst"]
lr = 1e-5

for dataset in datasets:
    print(f"Treinando o modelo pré-treinado para {dataset.upper()}...")
    args = set_args(dataset, option, lr)
    train(args)

    print(f"Avaliando o modelo pré-treinado para {dataset.upper()}...")
    test(args)
```
